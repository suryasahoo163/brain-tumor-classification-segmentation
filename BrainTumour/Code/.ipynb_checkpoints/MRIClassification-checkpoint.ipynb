### Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys,os,joblib,cv2
import keras
import sklearn
from sklearn.preprocessing import LabelEncoder
import pickle
```

### Importing the images

```python
image_loc = (r"C:\Users\surya_hwzjtax\Desktop\datascience\BrainTumour\MRI Image Classification")
dir_img = os.listdir(image_loc)

# loading training data
train_image = []
train_label = []

for dir in os.listdir(os.path.join(image_loc,dir_img[1])):
    for imagePath in os.listdir(os.path.join(image_loc,dir_img[1],dir)):
        train_image.append(cv2.resize(cv2.imread(os.path.join(image_loc,dir_img[1],dir,imagePath)),(256,256)))
        train_label.append(str(dir))

# loading testing data
test_image = [] 
test_label = []

for dir in os.listdir(os.path.join(image_loc,dir_img[0])):
    for imagePath in os.listdir(os.path.join(image_loc,dir_img[0],dir)):
        test_image.append(cv2.resize(cv2.imread(os.path.join(image_loc,dir_img[0],dir,imagePath)),(256,256)))
        test_label.append(str(dir))

```

### Image Preprocessing

```python
def preprocessing(image):
    image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
    image = image.reshape((256,256,1))
    return image
```

```python
for i in range(len(train_image)):
    train_image[i] = preprocessing(train_image[i])
for i in range(len(test_image)):
    test_image[i] = preprocessing(test_image[i])
```

### Label Encoding 

```python
encoder = LabelEncoder()
train_target = encoder.fit_transform(train_label)
test_target = encoder.transform(test_label)
```

```python
classes = list(encoder.classes_)
encoder_dict = {}
for i in classes:
    encoder_dict[encoder.transform([i])[0]] = (" ".join(i.split("_"))).title()
```

```python
encoder_dict
```

### Model Creation

```python
n_class = np.unique(np.array(train_label)).size
train_target = keras.utils.to_categorical(train_target,num_classes = n_class)
print(train_target.shape)
```

```python
test_target = keras.utils.to_categorical(test_target,num_classes = n_class)
print(test_target.shape)
```

```python
train_image = np.array(train_image)
test_image = np.array(test_image)
train_target = np.array(train_target)
test_target = np.array(test_target)

import tensorflow as tf
train_image = tf.cast(train_image, tf.int32)
train_target = tf.cast(train_target, tf.int32) #or appropriate type

train_image.shape
```

```python
model = keras.models.Sequential()

# Input layer

# Convolution layer
model.add(keras.layers.Conv2D(16,(3,3),activation="relu",input_shape=(256,256,1)))
# Dense layer
model.add(keras.layers.MaxPooling2D(pool_size=(2,2)))
# Convolution layer
model.add(keras.layers.Conv2D(8,(3,3),activation="relu"))
# Dense layer
model.add(keras.layers.MaxPooling2D(pool_size=(2,2)))
# Flatten
model.add(keras.layers.Flatten())

# Dense layer
model.add(keras.layers.Dense(units = 100,activation="relu"))
model.add(keras.layers.Dense(units = 100,activation="relu"))
model.add(keras.layers.Dense(units = n_class,activation="softmax"))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model_history = model.fit(train_image, train_label, epochs=30, batch_size=8)
```
