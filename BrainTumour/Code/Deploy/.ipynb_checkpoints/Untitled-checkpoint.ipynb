```python
from PIL import Image, ImageTk
import cv2
import numpy as np
import tkinter as tk

# Load PIL image
pil_image = Image.open(r"C:\Users\surya_hwzjtax\Desktop\datascience\BrainTumour\Code\Deploy\plus.png")

# PIL → OpenCV
cv2_image = cv2.cvtColor(np.array(pil_image), cv2.COLOR_RGB2BGR)

# (Optional OpenCV processing here)

# OpenCV → PIL
pil_image_again = Image.fromarray(cv2.cvtColor(cv2_image, cv2.COLOR_BGR2RGB))

# PIL → PhotoImage
photo_image = ImageTk.PhotoImage(pil_image_again)

# Display with Tkinter (example)
root = tk.Tk()
label = tk.Label(root, image=photo_image)
label.pack()
root.mainloop()
```
